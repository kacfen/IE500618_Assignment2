```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns
import numpy as np
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv("../../data/diabetes_binary_classification_data.csv")
```

```python
df.head(1)
```

```python
train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    random_state=13,
    stratify=df["Diabetes_binary"]
)

# Statify the data
X_train = train_df.drop(columns="Diabetes_binary")
y_train = train_df["Diabetes_binary"]

X_test = test_df.drop(columns="Diabetes_binary")
y_test = test_df["Diabetes_binary"]

print(train_df.shape, test_df.shape)
```

```python
binary = [
    "HighBP", "HighChol", "CholCheck", "Smoker", "Stroke",
    "HeartDiseaseorAttack", "PhysActivity", "Fruits", "Veggies",
    "HvyAlcoholConsump", "AnyHealthcare", "NoDocbcCost",
    "DiffWalk", "Sex"
]

numeric_ordered = [
    "BMI", "MentHlth", "PhysHlth",
    "GenHlth", "Age", "Education", "Income"
]

preprocess = ColumnTransformer([
    ("scale", StandardScaler(), numeric_ordered),
    ("binary", "passthrough", binary)
])

model = Pipeline([
    ("preprocess", preprocess),
    ("logistic", LogisticRegression(max_iter=1000))
])

model.fit(X_train, y_train)
```

```python
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import (
    confusion_matrix, ConfusionMatrixDisplay,
    accuracy_score, precision_score, recall_score, f1_score
)

cm = confusion_matrix(y_test, y_pred)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 score:", f1_score(y_test, y_pred))

ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No diabetes", "Prediabetes/diabetes"]
).plot()

plt.title("Confusion Matrix", fontweight="bold")
plt.show()
```
